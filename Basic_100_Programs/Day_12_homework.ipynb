Situation: A college canteen wants a small Python program to review the value of orders placed during one day. Assume the
order amounts are available to the program, for example: 120, 250, 80, 450, 175 and 300.
Your final program should be able to:
• Store and manage the order amounts.
• A new order of Rs. 200 arrives later and must be added.
• An order of Rs. 150 was missed and should be placed after the first order.
• One cancelled order should be removed from the data.
• Display the final list of order amounts.
• Calculate the total money collected.
• Calculate the average order value.
• Count how many final orders are Rs. 200 or more.
• Find the largest order amount without using a ready-made function that directly gives the maximum.
• Display a short end-of-day summary.

```python
List = [120,250,80,450,175,300]

print("Total order value for today :", List)

new_order = int(input("Enter the new order value ?"))
List.append(new_order)
print("Total order after new order :",List)

missed_order = int(input("Enter the missed order value?"))
List.insert(1,missed_order)
print("Total orders after inserting missed value :",List)

cancelled_order = int(input("Enter the cancelled order value :"))
List.remove(cancelled_order)
print("Total orders remainig after cancelled order:", List)

len_value = len(List)
#Total money for the day
sum = 0
count = 0
for i in List:
    sum = i+sum
    if i>200:
        count = count+1
    
print("Total money collected :",sum)
print("total order value greater than 200 is :", count)

average = sum/len_value
print("Average order value is :", average)

highest_order = max(List)
print("Largest order amount for the day is : ",highest_order)

print(f"End of day Summary : Total sale value is {sum},\naverage order value was {average}, \nTotal items that are greater than 200 is {count} \nwith highest order value is {highest_order}")



```
