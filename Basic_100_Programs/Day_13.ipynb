```python
#Take a sentence from the user and count its vowels.

sen = "Hi How are you?"#input("Enter a sentence")

count= 0

for i in sen:
    #print(i.upper())
    if i.upper() in 'AEIOU' :
        count = count+1
        #print("Match")
print('Total count of vowels :',count)


```

```python
#Check whether a given string is a palindrome (use slicing).

pal = 'madam'

#reverse
rev = pal[::-1]
if rev in pal:
    print("String is a Palindrome")
else:
    print("String is not a Plaindrome")
```

```python
#From a list of numbers, create a new list containing only the even numbers (use list comprehension).

l = [2,3,4,5,6,7,8,9,10]

r = [i for i in l if i%2 == 0]
r
```

```python
#Given a tuple of student marks, print the highest, lowest and average.

student = (34,56,7,8,89,35)

type(student)

highest = max(student)
lowest = min(student)
total = sum(student)
length = len(student)

print("Hoghest marks is :", highest)
print("Lowest marks is :", lowest)
print("Total marks obtained :",total)
print("Average is :",total/length)
```

```python
#Check if a number is even or odd using the bitwise & operator.

number = int(input("Enter a number :"))

if number & 1==0:
    print("Even")
else:
    print("Odd")


```

```python
#Count the frequency of each character in a string using a dictionary.

string_1 = "Hi how are you"

split_str = str_1.split()
print(split_str)
length_str = len(split_str)
#print(length_str)

for word in split_str:
    #print(word,":",split_str.count(word))
    for char in word:
        print(char,":",word.count(char))
```
