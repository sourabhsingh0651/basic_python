```python
#1. A company wants to calculate the number of possible arrangements of n different items using factorial

fact_number = int(input("Enter a number : "))

fact = 1

for i in range(1,fact_number+1):
    #print(i)
    fact = fact*i
print(fact)
```

```python
#2. Problem: A security system generates a verification value by multiplying all digits of a number. 
# Example: 423 -> 4 x 2 x 3 = 24 Another example: 5214 -> 40

digit = input("Enter digits greater than 10")

length = int(len(digit))
print("Digit Length is",length)

j =1
for i in digit:
    value = int(i)
    j = value*j 
print(j)
```

```python
#3. Ask the user for the number of terms n and generate the series:
# For n = 5, output: 2 6 12 20 30

limit = int(input("Enter the range :"))
add = 0
for i in range(1,limit+1):
    add = i*2 + add
    print(add)
```
