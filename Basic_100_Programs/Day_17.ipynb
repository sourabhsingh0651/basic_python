Homework 1 - Palindrome Number Read a positive integer and determine whether the number reads the same from left to right and right to left. Do not convert the number into a string. Example: 121 -> Palindrome, 123 -> Not Palindrome 

```python
num = 11211
original_num = num
i = 0
reverse = 0

while num>0:
    i = num%10
    reverse = reverse*10+i
    print("Reverse",reverse)
    #remove last digit
    num = num//10
    #print("Remaining",num)

if reverse == original_num:
    print(f"The number {original_num} is Palindrome")
else:
    print(f"The number {original_num} is not a Palindrome")

```

Homework 2 - Count the Factors Read a positive integer and display how many factors it has. Also display whether the number has more than two factors. Example: 12 has 6 factors 

```python
number = 5
count = 0
for i in range(1,13):
    #print(i)
    if number%i == 0:
        count = count+1

print(f"Total number of factors for the number: {number} is {count}")

if count>2:
    print(f"The number {number} has more than 2 factors")
elif count <=2:
    print(f"The number {number} has not more than 2 factorial")
```

Homework 3 - Perfect Number Check A perfect number is equal to the sum of its positive factors excluding the number itself. Create a program to determine whether an entered number is perfect. Example: 6 -> 1 + 2 + 3 = 6, therefore Perfect Number

```python
number = 28
sum_factors = 0
for i in range(1,number+1):
    
    if number%i == 0:
        print(f"the factorials are {i}")
        sum_factors = sum_factors+i
if sum_factors-number == number:
    print(f"The number {number} is a Prime Number")
else:
    print(f"the number {number} is not a Prime Number")

```
