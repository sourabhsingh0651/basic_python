```python
# 2. Create table(n, limit). It should display the multiplication table of n up to the given limit.

def table(n,limit):
    i = 1
    while i <= limit:
        print(n," x ",i, "=", n*i )
        i = i+1
print(table(5,8))
print(table(3,3))
print(table(4,4))
```

```python
# 3. Create factorial(n). The function should calculate and display the factorial of n.

def factorial(n):
    prod_var = 1
    fact = 1

    while prod_var <= n:
        #print(prod_var)
        fact = fact*prod_var
        prod_var = prod_var+1
    print(fact)

print(factorial(5))
```
