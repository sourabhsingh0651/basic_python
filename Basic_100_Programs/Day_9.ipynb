### One student has marks in five subjects. Store subject name and marks together using a nested list.

Your program should:
Use data similar to [["Python", 78], ["Math", 35], ...].
• Traverse the nested list using while.
• Display each subject name, marks and Pass/Fail status.
• Count how many subjects are passed and how many are failed.
• If all five subjects are passed, display "Overall: PASS"; otherwise display "Overall: NEEDS IMPROVEMENT".
Concepts you will practice: nested list, indexing, while, conditions, counters, logical thinking
Small hint: Subject name is data[i][0] and marks are data[i][1].

```python
student = [["Python", 78], ["Math", 35], ["English", 80], ["Hindi", 99], ["SST", 50]]
print("Initial List is : ",student)

len_student_list = len(student)
print("Length of List is :",len_student_list)

#loop
i = 0
count = 0
while i<len_student_list:
    subject = student[i][0]
    marks = student[i][1]
    #print(f"Subject is {subject} and marks obtained {marks}")
    if marks>40:
        print(f"Subject is {subject} and marks obtained {marks} : Pass")
        count = count+1
    else:
        print(f"Subject is {subject} and marks obtained {marks} : Fail")

    i = i+1
print("Total Pass Subjects :", count)

if count == 5:
    print("Overall: PASS")
else:
    print("Overall: NEEDS IMPROVEMENT")
```

### Situation: A small shop wants a reusable function to calculate a customer's bill.
Your program should:
Create calculate_bill(price, quantity=1) that returns price x quantity.
• Ask the user for price and quantity in the main program.
• Display the total bill.
• Also demonstrate one call where quantity is not provided, so the default value is used.
• Demonstrate one call using keyword arguments in reverse order.
• If price or quantity is not positive, display "Invalid Input" instead of calculating the bill.
Concepts you will practice: function with return, default parameter, keyword arguments, if, logical operators, arithmetic
Small hint: Validate the inputs before doing the multiplication.

```python
def calculate_bill(price, quantity=1):
    return price*quantity

#ask for price and quantity:
price = float(input("Enter the price of a product ?")) #price can be in float as well
quantity_value = input("Enter the quantity ?") #this can be null or int

if price < 0: #check for validation
    print("INVALID INPUT")

if quantity_value == "": #if null take default parameter
    quantity = 1
else:
    quantity = int(quantity_value) #if other convert to int

if price > 0 and quantity>0:
    result = calculate_bill(price,quantity)
    print("Total bill value is :",result)
```
