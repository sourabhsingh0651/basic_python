```python
l1 = [i*i for i in range(1,10) if i%2==0]

print(l1)
```

```python
#set comprehension:

s1 = {i*i for i in range(1,10)}

print(s1)
```

```python
#find dictinct len of list values
names = ["Ravi", "Amit", "Sara", "Neha", "Ram"]

len_set = {len(ele) for ele in names}

print(len_set)
```

```python
#dict comprehnsion

words = ["hi", "python", "set"]

lengths = {w:len(w) for w in words}

print(lengths)
```

```python
nums = [1,2,3,4,5,6]

even_sq = {n: n*n for n in nums if n%2==0}
print(even_sq)
```
