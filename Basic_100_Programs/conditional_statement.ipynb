```python
#Write a program to read a number and check whether it is even or odd.

int_num = int(input("Enter an integer value : "))

if int_num % 2 == 0:
    print("Even")

else:
    print("odd")
```

```python
#Write a program to read a number and check whether it is positive, negative or zero

int_number = int(input("Enter a number : "))

if int_number == 0:
    print("Number is 0")

elif '-' in str(int_number):
    print("Number is negative")

else:
    print("Number is positive")

```

```python
#Write a program to read three numbers and find the largest among them.

num_1 = int(input("Enter 1st number : "))
num_2 = int(input("Enter 2nd number : "))
num_3 = int(input("Enter 3rd number : "))

list_1 = [num_1, num_2, num_3]
print('Initial list is : ', list_1)

list_1.sort()
print(list_1)

#print last index
last_index = list_1[-1] 

print("last Index : ", last_index)
```

```python
#Write a program to read three numbers and find the smallest among them.

num_1 = int(input("Enter 1st number : "))
num_2 = int(input("Enter 2nd number : "))
num_3 = int(input("Enter 3rd number : "))

list_min = [num_1, num_2, num_3]

list_min.sort()

print("initial list :", list_min)

first_index = list_min[0]
print("Smallest number is : ", first_index)
```

```python
#Write a program to read a character and check whether it is an alphabet, digit or special symbol
import string
word = 'asdf23$#1'

#loop

for char in word:
    print(char)

    if char.isalpha():
        print("Its an alphabet")

    elif char.isnumeric():
        print("Its a numeric value")

    elif char in string.punctuation:
        print("Its a special char")
```

```python
#Write a program to read the marks of a student and print the grade (A/B/C/D/Fail).

total_marks = int(input("Enter the total marks out of 1000"))

if total_marks >= 900:
    print("Grade A")
elif total_marks >= 800:
    print("Grade B")
elif total_marks >= 700:
    print("Grade C")
elif total_marks >= 600:
    print("Grade D")
elif total_marks <=599:
    print("Fail")
    
```

```python
#Write a program to read a number and check whether it is divisible by both 3 and 5

number = int(input("Enter a number :"))

if number%3 == 0 and number%5 == 0:
    print("Divisible by 5 and 3")

else:
    print("Not divisible by 5 and 3")
```
