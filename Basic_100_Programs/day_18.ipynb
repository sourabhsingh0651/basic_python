```python
n = int(input("enter a prime number"))
i = 1
count = 0

while i<n:
    factor = n%i
    #print(factor)
    if factor == 0:
        count = count+1
        #print("Count Value : ",count)
    i=i+1
print("Count Value : ",count)

if count <= 2:
    print(f"Number {n} is a Prime number")
elif count > 2:
    print(f"Number {n} is not a prime number")
else:
    print(f"Number {n} is not a prime number")
```

```python

n = int(input("Enter a number: "))

i = 1
count = 0

while i <= n:
    factor = n % i

    if factor == 0:
        count = count + 1

    i = i + 1

print("Count Value:", count)

if n >= 2 and count == 2:
    print(f"Number {n} is a Prime number")
else:
    print(f"Number {n} is not a Prime number")
```

```python
n = int(input("Enter a number: "))

count = 0

while n > 0:
    last_number = n % 10
    count = count * 10 + last_number
    n = n // 10

print("Reversed number:", count)

```
