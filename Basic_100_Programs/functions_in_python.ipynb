```python
items = [1,2,2,3,4,4,4,5,9,9]

dup = []

for i in items:
    count = 0
    for j in items:
        if i == j:
            count = count+1
    if count>1 and i not in dup:
        dup.append(i)
print(dup)

```

```python
#define a method to returb aum and avg of a list

def avg_sum(num):
    return sum(num),sum(num)/len(num)

l1 = [12,38,10,40]
sum,avg= avg_sum(l1)
print("Sum is ",sum)
print("Avg is ", avg)
```

### *args and *kwargs

```python
# *args

def add(*args):
    sum = 0
    for ele in args:
        sum = sum+ele
    return sum

print(add(5,4,3))
```

```python
# *kwargs

def intro(**kwargs):
    return kwargs

intro(age=14,name="Sourabh")
```
