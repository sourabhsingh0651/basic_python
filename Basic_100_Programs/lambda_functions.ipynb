Map Function: High order function:
1.Call function for each element of collection.
2.Store the output or return object to a map object



```python
nums = [3,4,2,5,6,7]

sqr = list(map(lambda n:n**2,nums)) #[9,16,4,25,36,49]

print(sqr)
```

filter() - keeps only matching element
filter(function,collection)

```python
evens = list(filter(lambda n:n%2==0,nums))
print(evens)
```

```python
from functools import reduce

nums = [2,3,4,5]

prod = reduce(lambda a,b:a*b,nums)

print(prod)
```

```python
raw = ["$100", "$250", "$80"]

prices = list(map(lambda p: int(p.replace("$","")), raw))

print(prices)
```

```python
nums = [1,2,3,4,5]

double = list(map(lambda n:n*2,nums))

print(double)
```

```python
nums = [-3,5,-6,3,-1]

positive = list(filter(lambda n:n>=0,nums))

print(positive)
```
