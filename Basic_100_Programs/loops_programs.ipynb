```python
#Write a program to count how many numbers from 1 to n are divisible by 3.

number = int(input("Enter a number : "))

#display a range
for i in range(1,number+1):
    if i%3 == 0:
        print(f"{i} is divisible by 3.")
```

```python
#Write a program to display all numbers from 1 to n that are divisible by 3 or 5.

number = int(input("Enter a number : "))

#display a range
for i in range(1,number+1):
    print(i)

    if i%3 == 0:
        print(f"{i} is divisible by 3")

    elif i%5 == 0:
        print(f"{i} is divisible by 5")
```

```python
# Guessing game:

import random

random.randint(1,100) # generate random number

jackpot = random.randint(1,100)
print("Hidden Jackpot Number ", jackpot)

guess = int(input("Guess the number"))
counter_value = 1

while guess != jackpot:
    if guess<jackpot:
        print("Guess Higher")
    else:
        print("Guess Lower")

    guess = int(input("Guess Again Please"))
    counter_value = counter_value+1

print("Correct Guess")
print("You took ", counter_value , "attempts")

```
