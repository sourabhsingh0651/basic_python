Situation: You are creating a simple registration system for a website. The user must enter a username and password, and
the program should decide whether registration can continue.
Your program should:
Username rules:
• Must not contain spaces.
• Must contain at least 5 characters.
• Must not start with an underscore (_).
• Must not contain the word 'admin'.
Password rules:
• Must contain at least 8 characters.
• Must contain at least one uppercase letter.
• Must contain at least one lowercase letter.
• Must contain at least one number.
• Must contain at least one special character from @ # $ %.
If both username and password satisfy all rules, display 'Registration Successful'. Otherwise, display suitable messages explaining what
is wrong.

```python
def check_username(username):
    errors = []
    if " " in username:
        errors.append("User name must not contain spaces")
    if len(username) < 5:
        errors.append("User name must contain atleast 5 character")
    if username[0:1] == "_":
        errors.append("User name must not starts with _")
    if "admin" in username:
        errors.append("User name must not contain the word admin")
    return errors

def check_password(password):
    errors = []
    if len(password) < 8:
        errors.append("Password must contain 8 characters")
    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False
    special_chars = "@#$%"
    #lets look at each char one by one:
    for ch in password:
        if ch.isupper():
            has_upper = True
        elif ch.islower():
            has_lower = True
        elif ch.isdigit():
            has_digit=True
        elif ch in special_chars:
            has_special = True
    if not has_upper:
        errors.append("Password must contain atleat one uppercase letter")
    if not has_lower:
        errors.append("Password must contain atleat one lowercase letter")
    if not has_digit:
        errors.append("Password must contain atleat one digit")
    if not has_special:
        errors.append("Password must contain atleat one special letter")

    return errors



username = input("Enter username ?")
password = input("Enter PW")

if len(check_username(username)) == 0 and len(check_password(password)) == 0:
    print("You are Logged in")
else:
    print("Not Logged in")
```
