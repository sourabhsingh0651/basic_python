```python
#Write a program to count the number of digits in a number n.

number = int(input("Enter a number : "))

#display range
for i in range(1,number+1):
    digit_len = len(str(i)) # len always works on string not with the integers convert int to string for len.
    print(digit_len)
```

```python
# Write a program to find the sum of all digits of a number n.

number = '371989'

sum = 0

for i in number:
    i = int(i)
    #print(type(i)) ### type() ---> to check the data type
    #print(i)

    sum = sum+i

    print(sum)




```

```python
#Write a program to find the product of all digits of a number n.

number = '371989'

prod = 1

for i in number:
    i = int(i)
    prod = prod*i

    print(prod)
```

```python
#Write a program to reverse a number n.

number = 371989

number = str(number)

list_no = []

for i in number:
    #print(i)
    list_no.append(i)

#print(list_no)

list_no = list_no[::-1]

print(list_no)

```

```python
#Write a program to reverse a number n.

number = 133

reverse = 0

while number>0: ###133->13->1
    n = number%10 #--> gives 3->3->1
    reverse = reverse * 10 + n #---> 3,3,1
    number = number//10 ### ---> 13->1

print(reverse)

```
