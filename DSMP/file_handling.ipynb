Some Theory
Types of data used for I/O:
Text - '12345' as a sequence of unicode chars
Binary - 12345 as a sequence of bytes of its binary equivalent
Hence there are 2 file types to deal with
Text files - All program files are text files
Binary Files - Images,music,video,exe files


How File I/O is done in most programming languages
Open a file
Read/Write data
Close the file


```python
#writing to a file:

f = open('sample.txt','w')
f.write("Hello World")
f.close()
```

```python
#write multiple lines:

f = open('sample1.txt','w')
f.write("Hello world")
f.write("\nHow are you?")
f.close()
```

```python
# id the fille is already present:

f = open('sample.txt','w')
f.write("Salman Khan")
f.close()
```

```python
#how open() works:

f = open('sample1.txt','a')
f.write("\nI am fine")
f.close()
```

```python
#write lines:

L = ["hello\n", "hi\n", "how\n", "are\n"]

f = open('sample1.txt','a')
f.writelines(L)
f.close()
```

```python
#reading from files:

f = open("sample1.txt","r")
s = f.read()
print(s)
f.close()
```

```python
#reading upto n char

f = open("sample1.txt","r")
s = f.read()
print(s)
f.close()
```

```python
#read line by line:
f = open("sample1.txt","r")
print(f.readline(),end='')
print(f.readline(),end='')
f.close()
```

```python
f = open('sample1.txt','r')

while True:
    data = f.readline()

    if data == '':
        break
    else:
        print(data,end='')

f.close()

```

```python
#Using context manager (with)

#with

with open('sample1.txt','w') as f:
    f.write('selmon bhai')

```

```python
# read now:

with open('sample.txt','r') as f:
    print(f.read())
```

```python
# moving within a file

with open('sample.txt','r') as f:
    print(f.read(10))
    print(f.read(10))
```

```python
big_L = ["hello world" for i in range(1000)]

with open("big.txt", 'w') as f:
    f.writelines(big_L)

#chunking:

with open('big.txt', 'r') as f:
    chunk_size = 10

    while len(f.read(chunk_size)) > 0:
        print(f.read(chunk_size),end = '')
        f.read(chunk_size)
```

```python
#seek and tell

with open('sample.txt','r') as f:
    print(f.read(10))
    print(f.tell())
    print(f.seek(0))
    print(f.read(10))
```

```python
#binary files

with open("screenshot.png",'rb') as f:
    with open("screenshot-copy.png", 'wb') as wf:
        wf.write(f.read())
```
