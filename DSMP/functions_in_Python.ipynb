### args and kwargs
---> args and kwargs are special python keywords that are used to pass the variable lenghth of arguments to a function:


```python
### *args
# allows us to pass a variable number of non-keyword arguments to a function:

def multiply(*args):
    product = 1

    for i in args:
        product = product*i
    return product

multiply(1,2,3,4,5,6)
```

```python
###**kwargs

# key-value pair, like a dictionary

def display(**kwargs):

    for (key,value) in kwargs.items():
        print(key,"-->",value)

display(india='delhi',srilanka='colombo',nepal='kathmandu')
```

```python
def sort_sequence(seq):
    temp = []

    for i in sorted(seq.split('-')):
        temp.append(i)

    return "-".join(temp)

s = "green-red-yellow-black-white"
sort_sequence(s)
```

```python
# 3.  Write a Python function that accepts a string and calculate the number of upper case letters and lower case letters.

#str_1 = 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
def char_count(str_1):
    upper_count = 0
    lower_count = 0

    for char in str_1:
        if char.isupper():
            upper_count = upper_count+1
        elif char.islower():
            lower_count = lower_count+1

    return upper_count,lower_count
    

str_1 = 'CampusX is an Online Mentorship Program fOr EnginEering studentS.'
char_count(str_1)
```

Problem 5: Write a Python function to check whether a number is perfect or not.
A Perfect number is a number that is half the sum of all of its positive divisors (including itself).

Example :

The first perfect number is 6, because 1, 2, and 3 are its proper positive divisors, and 1 + 2 + 3 = 6. 
Equivalently, the number 6 is equal to half the sum of all its positive divisors: ( 1 + 2 + 3 + 6 ) / 2 = 6. 

The next perfect number is 28 = 1 + 2 + 4 + 7 + 14. This is followed by the perfect numbers 496 and 8128.

```python
def perfect_number(num):

    sum = 0

    for i in range(1,num):
        if num % i == 0:
            sum = sum+i

    return sum == num
perfect_number(3)

```

```python
#6. Write a Python function to concatenate any no of dictionaries to create a new one.

def merge_dict(*kwargs):
    d = {}

    for i in kwargs:
        d.update(i)

    return d

dic1={1:10, 2:20}
dic2={3:30, 4:40}
dic3={5:50,6:60}

merge_dict(dic1,dic2,dic3)
```

```python
#7. Write a python function that accepts a string as input and returns the word with most occurence.

#Input:
# hello how are you i am fine thank you

def most_used(s):
    d = {}
    for i in s.split():
        if i in d:
            d[i] = d[i] + 1
        else:
            d[i] = 1

    max_value = max(d.values())

    for i in d:
        if d[i] == max_value:
            print(i,'-->',d[i])
            break

most_used("hello how are you i am fine thank you")
```

```python
# 8. Write a python function that receives a list of integers and prints out a histogram of bin size 10
import math

def histogram(L):

    min_bin = math.floor(min(L)/10)*10
    max_bin = math.ceil(max(L)/10)*10

    d={}

    for i in range(min_bin,max_bin,10):
        #print(i+1,i+10)
        count = 0
        for j in L:
            if i+1<=j<=i+10:
                count = count+1
        d[str(i+1) + '-' + str(i+10)] = count

    return d

histogram(L = [13,42,15,37,22,39,41,50])
```
