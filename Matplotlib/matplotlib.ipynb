```python
### groupby() ---> 

data = df.groupby('industry')['total_laid_off'].sum()

data
```

```python
x = data.index
y = data.values

plt.figure(figsize=(10,5))
plt.xticks(rotation=90)
plt.title('Industry wise layoffs')
plt.xlabel('Industry')
plt.ylabel('Number of Layoffs')
plt.plot(x,y,color='red',linestyle='--',linewidth=1.5,marker='o',markersize='7')
plt.grid()
plt.show()

```
