### Situation: You want one reusable function that can print different multiplication tables from different Python files.
Your program should:
Create table_tools.py with a function table(n, limit=10).
Create main.py and import the file as a short alias using import table_tools as t.
Call it to display:
• table of 5 using the default limit,
• table of 7 only up to 5,
• one call using keyword arguments in a different order, such as t.table(limit=4, n=12).
Concepts you will practice: functions, parameters, default parameter, keyword arguments, while, module alias
Small hint: The table function should contain the loop. main.py should mainly call the function.
