```python
### Stacking

###horizontal stacking ---> row by row stacking -> hstack
###vertical stacking ---> col by col ->vstack()
### size of the array must be the same

array_1 = np.arange(1,13).reshape(3,4)

array_1
```

```python
array_2 = np.arange(13,25).reshape(3,4)

array_2
```

```python
horizontal_array = np.hstack((array_1,array_2))

horizontal_array
```

```python
vertical_array = np.vstack((array_1,array_2))

vertical_array
```

```python
### split - hsplit()--->vsplit()

np.hsplit(vertical_array,2)
```

```python
np.vsplit(vertical_array,2)
```
