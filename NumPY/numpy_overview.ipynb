#### Numpy Arrays vs Python Sequences:

- Numpy Arrays have a fixed size at creation, unlike python lists which grows dynamicallly.
- Changing the size of n dimesnsional array will create the new array but delete the original one.
- The elements in numpy array are required to be of the same data type.
- Thus will be of the same size in memory.

### Numpy Array facilitates advanced mathematical and other opertaions on large number of data. typically such operations execute more efficiently with less code than with comparison in python built in functions.
