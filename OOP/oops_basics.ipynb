```python
class Abc:
    
    def __init__(self,x,y):
        self.a=x
        self.b=y
        print("Object Created")

    def add(self):
        print(f"Sum = {self.a+self.b}")

    def prod(self):
        print(f"Product = {self.a*self.b}")

obj = Abc(10,20)
obj1 = Abc(30,40)
obj.add()
obj.prod()
```
