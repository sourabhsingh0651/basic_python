### Data Frames Attributes and Functionality

- Attributes:
1. shape ---> provides the info about the row,column information about dataset --> df.shape
2. dtypes ---> It provides the idea about data types of all columns with their name and their associated data type will be shown by using dtypes. --> df. dtypes
3. index --->  for a data frame with labelled index we can get the idea about the label by just adding labelled index. ---> df.index
4. columns ---> it provides us the idea about all of the column names of the given data frames. ---> df.columns
5. values ---> it provides us the idea about the values present row by row inside the data frame. always returns numpy array representing the values of column row by row. ---> df.values
6. ndim ---> it provides us the idea about the dimension of the dataframes. df.ndim
7. size ---> provides us the idea of total number of elements present inside the data frames. All row*all columns.

```python
#strip() ---> helps to remove any additional spaces of brackets

ipl_batsman_df['Highest Score'] = ipl_batsman_df['HS'].str.strip("*")
ipl_batsman_df
```

```python
### adding columns

ipl_batsman_df["Lowest Score"] = 0
ipl_batsman_df
```

```python
ipl_batsman_df['Total Score'] = str(ipl_batsman_df['Highest Score']) + str(ipl_batsman_df['Lowest Score'])
ipl_batsman_df
```

```python
#df.drop() ---> removes the column
ipl_batsman_df.drop('Total Score')
ipl_batsman_df
```
