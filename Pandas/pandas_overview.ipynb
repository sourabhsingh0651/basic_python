### Pandas series is a column in a table like in excel or google sheets or labels called as indexes attached to each values.

- Indexes are going to be labelled.
-

```python
import pandas as pd
import numpy as np

pd.Series([40,50,60],index=['Bread','Milk','Butter'])
```

### Data Type ---> Integers ---> Data in series(column) ----> Qunatity of items in the stock.
### Data type ---> Float ---> Data in series (column) ----> Price of product ratings
### Data type ---> Strings ---> Data in series (column) ----> Product Name
### Data type ---> Bool ---> Data in series (column) 
### Data type ---> Date/Times ---> Data in series (column) 

```python
### ways to create a series in multiple ways:

### 1. List
### 2. NumPy Array
### 3. Dictionary
### 4. Custom Index
### 5. Scalar Value
```

```python
# List

country = ['India', 'China', 'USA', 'Pakistan']


pd.Series(country)
```

```python
batsman_runs = [48,34,51,100,187]

pd.Series(batsman_runs,dtype='int32')
```

```python
subject_names = ['maths','physics','bio','english']
subject_marks = [48,55,60,70]

pd.Series(data=subject_marks,index=subject_names,name='subject marks data')
```

```python
# Create series from numpy array...

values = np.arange(10,21)
pd.Series(values)
```

```python
### Sries from a dictionary....

data_items = {'Maths':78,'English':92, 'Physics':88, 'Chem':98}

pd.Series(data_items)
```

```python
### Custome Index...
values = [1,2,3,4,5]

pd.Series(data=values,index=['a','b','c','d','e'])
```

```python
#### Scalar value....

pd.Series(4,index=['a','b','c','d','e'])
```
