### LIST METHODS

1. append() - insert items at the end of a list
2. insert() - insert the value at a specific location/index
3. remove() - remove the value from our list that we are going to pass
4. pop() - remove the value based on index
5. index() - provides the index of the value that we are going to pass inside it if present
6. sort() - sorting the values in ascending manner
7. reverse() - reverse the list structure
8. count() - counts the number of occurance of a specific value that we are passing inside in you list
9. clear() - clears the list.

```python
number_list = [11,24,12,8,6,9,7]
```

ENUMERATE FUNCTION ()

helps us to get the index value along with the actual value alomg with the combination that we are storing inside the list

```python
for index_value,value in enumerate(number_list):
    print(index_value, value)
```

Len () - Length function gives us the lenght of the list -  as total number of items we are storing inside the list.



```python
#len()

length = len(number_list)
print(length)
```

List Comprehension 

```python
#create a list storing values from 1 to 100

# [expression for item in iterable if condition]
# expression -  what we want to print
# for item in iterable - looping condition
# if condition - conditional statement

# Store value from 1-50 inside a list

list_items = [x for x in range(1,51)]
list_items







```

```python
list_squared = [items**2 for items in range(1,16)]
print(list_squared)
```

```python
list_product = [y for y in range(10,20)]
print(list_product)
```

```python
even_list = [x**2 for x in range(1,19) if x%2==0] # list with squares
even_list
```
