- Store the values within the curly {}.
- And are known for storing the values which are non duplicates.
 

```python
#Create a set
set_value={1,2,3,4,5}

type(set_value)
```

### Method in set

1. add() - insert a value in set function


```python
set_value.add(7)
set_value
```

```python
fruits = {'apple', 'banana', 'garapes', 'guava', 'pineapple'}

fruit_vegetables = ['apple', 'banana', 'potato', 'onion', 'capsicum']

for item in fruit_vegetables:
    fruits.add(item)

fruits
```
