### Nested Conditional Statements 

- involves if statements inside another if statements

```python
# Compare the height of 3 friends

Amit = int(input("Enter you height : "))
Prabhat = int(input("Enter you height : "))
Sourabh = int(input("Enter your height : "))

if Amit > Prabhat:
    if Prabhat > Sourabh or Amit > Sourabh:
        print("Amit is taller")

elif Prabhat > Amit:
    if Prabhat > Sourabh or Amit > Sourabh:
        print("Prabhat is Tallest")

else:
    print("Sourabh is Tallest")
```

```python
# Leap Year

year = int(input("Enter the year value : "))

if year%4 == 0:
    if year%100 == 0:
        if year%400 == 0:
            print("The current year is a leap year")

        else: 
            print("The current year is not a leap year")

    else:        
        print("Year is a leap year")

else:
    print("year is not a leap year")
```

```python
year = int(input("Enter the year value: "))

if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("The current year is a leap year")
else:
    print("The current year is not a leap year")
```
