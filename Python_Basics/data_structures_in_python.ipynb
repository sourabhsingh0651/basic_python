```python
fruit_list = ["apple", "banana", "orange", "guava", "pineapple", "grapes"]

print("Total length of items : ",len(fruit_list))

#+ve indexing
print(fruit_list[1])
print(fruit_list[0])
print(fruit_list[5])

#-ve indexing

print(fruit_list[-1])
print(fruit_list[-2])
print(fruit_list[-5])
```

### SLICING - read the contents from a certain position to another position

[:] ---> using colons

LIST SLICING CAN HELP US OUT WITH RETRIEVING A SPECIFIC DATA POINTS FROM OUR LIST.

```python
print(fruit_list[:])

#How it works

fruit_list[1:4] # Give the start index and index - 1 to the elemnt you want to fetch

fruit_list[2:6]

fruit_list[:6]

fruit_list[3:]
```

### Reverse the List

USE :: colon with -1

[::-1]

```python
fruit_list[:]

fruit_list[::-1]
```

```python
veg_list = ["Tomato", "Brinjal", "Potato", "Capsicum", "Onion"]

veg_list[:]

veg_list[::-1]

veg_list[1:5]

print("Reverse the list : ",veg_list[::-1])

```

### List are Mutable

By mentioning that list are mutable we mean to say that we can replace the values present at specific location inside opur list.



```python
#Replace the values

veg_list[:]

print(veg_list[:])

veg_list[2]

veg_list[2]="Apple"

print(veg_list[:])

veg_list[:]

veg_list[1:3]= "Banana","grape"

veg_list
```

### Use Loop to iterate

```python
# FOR Loop

veg_list = ["Tomato", "Brinjal", "Potato", "Capsicum", "Onion"]

for item in veg_list:
    print(item)

```
