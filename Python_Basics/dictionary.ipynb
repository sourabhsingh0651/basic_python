```python
### Nested Dictionary -> to keep a dict into another dict

nested_dict = {
    "Student_1" : {"Name" : "Amit", "Marks" : 84, "Grade" : "A"},
    "Student_2" : {"Name" : "Sourabh", "Marks" : 90, "Grade" : "A+"},
    "Student_3" : {"Name" : "Vikash", "Marks" : 56, "Grade" : "B"}
    }
```

```python
nested_dict
```

```python
# Iterate over nested dictiobaries

for outer_key,outer_value in nested_dict.items():
    print(f"Outer key is :{outer_key} and outer value is : {outer_value} ")

    for key_1,value_1 in outer_value.items():
        print(f"Inner key is : {key_1} and inner value is : {value_1}")
```

# DICTIONARY COMPREHENSION

Looping

Syntax : {expression for items in iterable if condition}

```python
comprehensive = {x**2 for x in range(11)}

comprehensive
```

```python
comprehensive = {x:x**2 for x in range(11)}
comprehensive
```

```python
even_dictionary={x:x**2 for x in range(1,11) if x%2==0}

even_dictionary
```

### MERGED DICTIONARIES



```python
dictionary_1 = {'a' : 1, 'b' : 2, 'c' : 3}
dictionary_2 = {'a' : 1, 'b' : 4, 'c' : 7}

merged_dictionary = {**dictionary_1,**dictionary_2}
merged_dictionary
```
