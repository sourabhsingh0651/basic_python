```python
def translator(English_text):
    """ This function i going to be used for converting english language to hindi language """
    if English_text == "hello" or English_text == "Hello":
        print(f"The translation for {English_text} is Namaste")

    if English_text == "help" or English_text == "Help":
        print(f"The translation for {English_text} is Madad")

    if English_text == "require" or English_text == "Require":
        print(f"The translation for {English_text} is Aavshyak")
```

```python
# Call the function - mention the name of the function and provide the input

input_value = input("Enter your english word : ")

translator(input_value)
```

```python
## function to convert English to hindi:

def translator_return(English_text):
    """ This function i going to be used for converting english language to hindi language """
    if English_text == "hello" or English_text == "Hello":
        Hindi_word = "Namaste"

    if English_text == "help" or English_text == "Help":
        Hindi_word = "Madad"

    if English_text == "require" or English_text == "Require":
        Hindi_word = "Aavshyak"

    return Hindi_word
```

```python
English_return_input = input("Enter your english word to be translated to hindi : ")

Hindi_Output = translator_return(English_return_input)
print(Hindi_Output)
```

```python
### Scenario 1: which is going to take input number from the user side and will figure out wether if the number is even or odd.

def even_or_odd(num):

    if num%2 == 0:
        output = "Even number"

    else:
        output = "Odd number"

    return output
```

```python
print(even_or_odd(5))
```

```python
print(even_or_odd(996))
```

```python
# Handling the errors in functions based on the input

def even_or_odd_instance(num):

    if isinstance(num,int):
        if num %2 == 0:
            print(f"{num} is even")

        else:
            print(f"{num} is odd")

    elif isinstance(num,str):
        print("Invalid Data Type")

    else:

        print(f"{num} is an invalid value for this")

```

```python
even_or_odd(14.4)
```

### Recursive Functions : By a recursive function we can say that a function is going to call itself again and again.



```python
## Factorial of a number : 8 -> 8*7*6*5*4*3*2*1

def factorial(num):
    if num==0 or num==1:
        print(f'The factorial of {num} is 1')

    else:
        factorial=1
        for i in range(1,num+1):
            factorial= factorial*i

        print(f"The factorila of num is : {factorial}")
        
```

```python
number = int(input("Enter the number : "))
factorial(number)
```
