### # range () -  function - to iterate through a series of numbers

### range(initial value, final value+1, step size(optional))

```python
### range function can be used for working as our iterable

### range() - takes an input as a parameter and the firts parameter taht the range function will be taking as the input the initial value that we want to print
### or we want to opearate on

# Print the numbers from 1-50

for item in range(1,11,1): ### range(initial value, final value+1, step size(optional))
    print(item)
```

### In the range function there is never a necessity mentioning the step size or gap.

- If we want to print the numbers or if we want to operate on numbers again there is no necessity for passing the initial value as well.
- If only my initial value is 0.

```python
# Iterate from 0-100 

for i in range(101):
    print(i)
```

```python
# Print the numbers from 0-100n with a gap of 5

for i in range(0,101,5):
    print(i)
```

### Now if we want to decrement the range.


```python
# Print from 50 - 1 in decrementing manner

for i in range(50,0,-1):
    print(i)
```

```python
# print the numbers from 50 - 0 with a gap of 5

for i in range(50,-1,-5):
    print(i)
```

### WHILE LOOP:

- while loop is being used for iterating over a series of numbers when a condition is true.
- And we have to initialize the value as well while iterating.

```python
### by making the use of while loop we need to print the number 0-50
i=0 #initialize

while i<=50: #condition
    print(i)
    i=i+1 #iteration
```

```python
# print only the even numbers:

for i in range(51):
    if i%2 == 0:
        print(i)
```

```python
# Print even numbers using while loop

i = 0

while i <= 50:
    if i%2 == 0:
        print(i)
    else:
        pass #nothing to be done or no action
    i = i+1 # always increment to avoid infinite loop
```

### Loop control statements:

1. break()
2. continue()

- Used to terminate a loop in between a condition is met.

```python
# break() ---> to terminate the loop when a condition is met.

# Print the numbers from 0-50 but once if 18 is acieved terminate the loop - do not print 18 as well.

for i in range(51):
    if i == 18:
        break
    else:
        print(i)
```

```python
# continue ---> which helps us in iterating over a series of numbers but helps us with skipping the values in between once we meet the condition.

# print the numbers from 0-50 but only print even numbers an for odd numbers skip the values.

for i in range(51):
    if i%2 == 0:
        print(i)

    else:
        continue
```

```python
### print the numbers fom 0-20 but once 17 is occuring skip it.

for i in range(21):
    
    if i == 17:
        continue
    else:
        print(i)
```

### Pass Statement - when you dont want to do anything

```python
# print the numbers 0 - 20 once 17 appears do nothing

for i in range(21):
    if i==17:
        pass
    else:
        print(i)
```
