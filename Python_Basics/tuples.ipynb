- Tuples are second type of DS which can store the data in a ordered way.
- Just like alist but they are not mutable
- () ---> Create using brackets

```python
#create tuple

a = tuple()
type(a)
```

```python
tuple_values = (1,2,3,4,5)
print(tuple_values)
```

```python
#tuples inside the nested list
tuple_list = [(1,2,3),(4,5,6),(11,12,14,15)]

tuple_list[2][:]
```

```python
#Nested Tuple

nested_tuple = ((1,2,3),(4,5,6,),(11,12,13,14))

nested_tuple[1][:]
```

```python
#iterate over a tuple

for i in nested_tuple:
    for j in i:
        print(j, sep=" ")
    print(" ")    
```
